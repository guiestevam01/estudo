```python
import getpass
import os

if "HUGGINGFACEHUB_API_TOKEN" not in os.environ:
    os.environ["HUGGINGFACEHUB_API_TOKEN"] = getpass.getpass("Cole seu token do Hugging Face aqui: ")

from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEndpointEmbeddings
from langchain_huggingface import HuggingFaceEndpoint
from langchain_huggingface import ChatHuggingFace
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough


embeddings_model = HuggingFaceEndpointEmbeddings(model="BAAI/bge-small-en-v1.5")

# Usando o Qwen 2.5 Instruct que possui suporte oficial a chat na API do HF
llm_base = HuggingFaceEndpoint(
    repo_id="Qwen/Qwen2.5-7B-Instruct",
    temperature=0.1,
    model_kwargs={"max_tokens": 200}
)
llm = ChatHuggingFace(llm=llm_base)
```

```python
pdf_link = "1758887557527-attachment.pdf"

loader = PyPDFLoader(pdf_link, extract_images=False)
pages = loader.load_and_split() #carrega em paginas
```

```python
#separar em chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 4000, #tamanho
    chunk_overlap = 20, # primeiro trecho = 1 chunk, o meu segundo chunk quero que tenha um pequeno overlap com anterior para nao ter perca de conteudo
    length_function = len,
    add_start_index = True
)
chunks = text_splitter.split_documents(pages)
```

```python
#Salvar no VectorDB - Chroma

#salva dentro da passa persistencia
db = Chroma.from_documents(chunks, embedding=embeddings_model, persist_directory="text_index")
db.persist()
```

```python
#carregar
vectordb = Chroma(persist_directory="text_index",embedding_function=embeddings_model)

#load retrievier
retriever = vectordb.as_retriever(search_kwargs={"k": 3}) # traz 3 doc relevantes
#construcao da cadeia do prompt para chamada do llm
chain = load_qa_chain(llm,chain_type="stuff")
```

```python
def ask(question):
    context = retriever.get_relevant_documents(question) #recupera os chunks relevantges a question
    answer = (chain({"input_documents": context, "question": question}, return_only_output=True))['output_text']
    return answer
```

```python
user_question = input("User": ")
answer = ask(user_question)
print("Answer: ", answer)
```
